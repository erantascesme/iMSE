# Reproduce the iMSE benchmark

1. From this checkout, install the package in this notebook's Python environment with `pip install -e .`.
2. Choose the dataset, device, and work directory below. Leave `BACKBONE` empty to run all its paper backbones.
3. Run the notebook. The default stage downloads data, extracts features, and evaluates the detectors. Use `evaluate` to reuse a completed feature cache.

For ImageNet-1K access and Vim-Base setup, see the [README](README.md).

```python
from pathlib import Path
import subprocess
import sys

REPO = Path.cwd()
if not (REPO / "imse_benchmark").is_dir():
    raise RuntimeError("Open this notebook from the iMSE repository root.")

DATASET = "CIFAR10"  # CIFAR10, CIFAR100, ImageNet1K
BACKBONE = ""  # e.g. ResNet18, or empty for all
STAGE = "all"  # all, download, extract, evaluate
DEVICE = "cuda"  # or cpu for supported backbones
WORK = REPO / "work"
```

```python
args = [sys.executable, "-m", "imse_benchmark", STAGE, "--dataset", DATASET,
        "--work-dir", str(WORK), "--device", DEVICE]
if BACKBONE:
    args += ["--backbone", BACKBONE]
subprocess.run(args, cwd=REPO, check=True)
```
